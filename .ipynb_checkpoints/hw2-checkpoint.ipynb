```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Perceptron
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.feature_extraction.text import CountVectorizer
```

```python
# perceptron function
def perceptron(p_tol, p_random_state, X_train, y_train, X_dev, y_dev, X_test, y_test):
    clf_p = Perceptron(tol=p_tol, random_state=p_random_state)
    
    # train the model
    clf_p.fit(X_train, y_train)
    
    #compare train accu and dev accu
    print("Train accuracy is: ", clf_p.score(X_train, y_train))
    print("Development accuracy is: ", clf_p.score(X_dev, y_dev))
    
    #test model
    print("Test accuracy is: ", clf_p.score(X_test, y_test))
    
    return clf_p
```

```python
#logistic regression function
def loss_reg(lr_max_iter, lr_random_state, X_train, y_train, X_dev, y_dev, X_test, y_test):
    clf_lr = LogisticRegression(max_iter=lr_max_iter, random_state=lr_random_state)
    
    #train model
    clf_lr.fit(X_train, y_train)
    
    #compare train accu and dev accu
    print("Train accuracy is: ", clf_lr.score(X_train, y_train))
    print("Development accuracy is: ", clf_lr.score(X_dev, y_dev))
    
    #test
    print("Test accuracy is: ", clf_lr.score(X_test, y_test))
    
    return clf_lr
```

```python
#SVM(LinearSVC) function
def svm(svm_random_state, svm_tol, ispipeline, X_train, y_train, X_dev, y_dev, X_test, y_test):
    if ispipeline: 
        clf_svm = make_pipeline(StandardScaler(), LinearSVC(random_state=svm_random_state, tol = svm_tol))
    else:
        clf_svm = LinearSVC(random_state=svm_random_state, tol = svm_tol)
    #train model
    clf_svm.fit(X_train, y_train)
    
    #compare train accu and dev accu
    print("Train accuracy is: ", clf_svm.score(X_train, y_train))
    print("Development accuracy is: ", clf_svm.score(X_dev, y_dev))
    
    #test
    print("Test accuracy is: ", clf_svm.score(X_test, y_test))

    return clf_svm
```

```python
#virtualization function
def visualization(results, table_title):   
    vir = results.plot(
        x="Model",
        kind="bar",
        figsize=(9, 5),
        ylim=(0, 1.05)
    )
    for container in vir.containers:
        vir.bar_label(container, fmt="%.3f", padding=3)
    plt.title(table_title)
    plt.xlabel("Linear Classification Model")
    plt.ylabel("Accuracy")
    plt.show()
    
```

```python
#------------- Spambase: distinguish between spam and non-spam emails -------------

#setup
#Split dataset into training, development, test sets
data_spambase = pd.read_csv("spambase/spambase.data", header=None)
X_spamb = data_spambase.iloc[:, :-1]
y_spamb = data_spambase.iloc[:, -1]

X_train_spamb, X_dev_test_spamb, y_train_spamb, y_dev_test_spamb = train_test_split(
    X_spamb, y_spamb, test_size=0.2, random_state=42
)

X_test_spamb, X_dev_spamb, y_test_spamb, y_dev_spamb = train_test_split(
    X_dev_test_spamb, y_dev_test_spamb, test_size = 0.5, random_state = 42
)

print(data_spambase.shape)
print(X_train_spamb.shape)
print(X_dev_spamb.shape)
print(X_test_spamb.shape)
```

```python
# Porceptron step

#1st try:
#para: tol=1e-3, random_state=0
#Train accuracy is:  0.6141304347826086
#Development accuracy is:  0.579175704989154


#2nd try: 
#para: tol=None, random_state=0
#Train accuracy is:  0.9051630434782608
#Development accuracy is:  0.9088937093275488
#Test accuracy is:  0.908695652173913
#accu is good enough. It seems no overfitting since the dev accu is similar to train accu

clf_p_spamb = perceptron(None, 0, X_train_spamb, y_train_spamb, X_dev_spamb, y_dev_spamb, X_test_spamb, y_test_spamb)
```

```python
# Loss Regression step

#1st try:
#para: max_iter=5000, random_state=0
#Train accuracy is:  0.9317934782608696
#Development accuracy is:  0.9219088937093276
#Test accuracy is:  0.9239130434782609
#accu is good enough. It seems there is no overfitting since the dev accu is similar to train accu
    
clf_lr_spamb = loss_reg(5000, 0, X_train_spamb, y_train_spamb, X_dev_spamb, y_dev_spamb, X_test_spamb, y_test_spamb)
```

```python
# SVM step

#1st try:
#para: random_state=0, tol = 1e-05
#Train accuracy is:  0.9296195652173913
#Development accuracy is:  0.9197396963123644
#Test accuracy is:  0.9195652173913044
#accu is good enough. It seems there is no overfitting since the dev accu is similar to train accu
   
clf_svm_spamb = svm(0, 1e-05, True, X_train_spamb, y_train_spamb, X_dev_spamb, y_dev_spamb, X_test_spamb, y_test_spamb)
```

```python
# Visualize the results of these 3 models

results_spamb= pd.DataFrame({
    "Model": ["Perceptron", "Logistic Regression", "LinearSVC"],
    "Train Accuracy": [clf_p_spamb.score(X_train_spamb, y_train_spamb),
                      clf_lr_spamb.score(X_train_spamb, y_train_spamb),
                      clf_svm_spamb.score(X_train_spamb, y_train_spamb)
                      ],
    "Development Accuracy": [clf_p_spamb.score(X_dev_spamb, y_dev_spamb),
                             clf_lr_spamb.score(X_dev_spamb, y_dev_spamb),
                             clf_svm_spamb.score(X_dev_spamb, y_dev_spamb)
                            ],
    "Test Accuracy": [clf_p_spamb.score(X_test_spamb, y_test_spamb),
                     clf_lr_spamb.score(X_test_spamb, y_test_spamb),
                     clf_svm_spamb.score(X_test_spamb, y_test_spamb)
                     ]
})

visualization(results_spamb, "Accuracy Comparison of Three Models for Spambase")
```

```python
#------------- Universal Declaration of Human Rights: distinguish between English and Dutch text -------------

#setup databases
with open("universal-declaration/english.txt", "r", encoding = "utf-8") as f_en:
    lines_en = [line.strip() for line in f_en.readlines()]
with open("universal-declaration/dutch.txt", "r", encoding = "utf-8") as f_nl:
    lines_nl = [line.strip() for line in f_nl.readlines()]
with open("universal-declaration/english_dev.txt", "r", encoding = "utf-8") as f_en_dev:
    lines_en_dev = [line.strip() for line in f_en_dev.readlines()]
with open("universal-declaration/dutch_dev.txt", "r", encoding = "utf-8") as f_nl_dev:
    lines_nl_dev = [line.strip() for line in f_nl_dev.readlines()]
with open("universal-declaration/english_test.txt", "r", encoding = "utf-8") as f_en_test:
    lines_en_test = [line.strip() for line in f_en_test.readlines()]
with open("universal-declaration/dutch_test.txt", "r", encoding = "utf-8") as f_nl_test:
    lines_nl_test = [line.strip() for line in f_nl_test.readlines()]

#train database
data_udhr_train = pd.DataFrame({
    "text": lines_en + lines_nl,
    "language": ["en"] * len(lines_en) + ["nl"] * len(lines_nl)
})

#dev database
data_udhr_dev = pd.DataFrame({
    "text": lines_en_dev + lines_nl_dev,
    "language": ["en"] * len(lines_en_dev) + ["nl"] * len(lines_nl_dev)
})

#test database
data_udhr_test = pd.DataFrame({
    "text": lines_en_test + lines_nl_test,
    "language": ["en"] * len(lines_en_test) + ["nl"] * len(lines_nl_test)
})

print(data_udhr_train.shape)
print(data_udhr_train.head())
print(data_udhr_dev.shape)
print(data_udhr_dev.head())
print(data_udhr_test.shape)
print(data_udhr_test.head())
```

```python
# set up for trian, dev, test set

X_train_text_udhr = data_udhr_train["text"]
y_train_udhr = data_udhr_train["language"]

X_dev_text_udhr = data_udhr_dev["text"]
y_dev_udhr = data_udhr_dev["language"]

X_test_text_udhr = data_udhr_test["text"]
y_test_udhr = data_udhr_test["language"]

print(X_train_text_udhr.shape)
print(X_dev_text_udhr.shape)
print(X_test_text_udhr.shape)
```

```python
# Text vectorization

#1st try: TF-IDF features 
#vectorizer = TfidfVectorizer()
#three models have the same accu, test and dev accu is 1, test accu is 0.975
#the accu are very high and wired


#2nd try: CountVectorizer
#para: analyzer='char', ngram_range=(1, 2)

#3rd try: CountVectorizer
#para: analyzer='char', ngram_range=(1, 3)

#4th try: CounterVectorizer
#para: analyzer='char', ngram_range=(2, 3)
vectorizer = CountVectorizer(analyzer='char', ngram_range=(2, 3))
X_train_udhr = vectorizer.fit_transform(X_train_text_udhr)
X_dev_udhr = vectorizer.transform(X_dev_text_udhr)
X_test_udhr = vectorizer.transform(X_test_text_udhr)

# note that y don't need to be vectorized
# sklearn can convert y from characters to suitable representation

print(X_train_udhr.shape)
print(X_train_udhr[:5].toarray())
print(X_dev_udhr.shape)
print(X_dev_udhr[:5].toarray())
print(X_test_udhr.shape)
print(X_test_udhr[:5].toarray())
```

```python
# Porceptron step

#1st try (TF-IDF features):
#para: tol=1e-3, random_state=0
#Train accuracy is:  1.0
#Development accuracy is:  1.0
#Test accuracy is:  0.975
#The accu is very high and need to think about if there is data leak or the test set is too small

#2nd try (CountVectorizer: ngram_range=(1, 2))
#para: tol=1e-3, random_state=0
#Train accuracy is:  1.0
#Development accuracy is:  0.975
#Test accuracy is:  0.9
#The train accu is very high, but the dev accu is better, test accu is ok

#3rd try (CountVectorizer: ngram_range=(1, 3))
#para: tol=1e-3, random_state=0
#Train accuracy is:  1.0
#Development accuracy is:  0.975
#Test accuracy is:  0.85
#The train accu is very high, but the dev accu is better, test accu is ok

#4th try (CountVectorizer: ngram_range=(1, 3))
#para: tol=1e-3, random_state=0
#Train accuracy is:  1.0
#Development accuracy is:  1.0
#Test accuracy is:  0.975
#The train accu is very high, but the dev accu is better, test accu is ok
clf_p_udhr = perceptron(1e-4, 0, X_train_udhr, y_train_udhr, X_dev_udhr, y_dev_udhr, X_test_udhr, y_test_udhr)
```

```python
# Loss Rwgression step

#1st try (TF-IDF features):
#para: max_iter=5000, random_state=0
#Train accuracy is:  1.0
#Development accuracy is:  1.0
#Test accuracy is:  0.975
#The accu is very high and need to think about if there is data leak or the test set is too small

#2nd try (CountVectorizer: ngram_range=(1, 2)))
#para: max_iter=5000, random_state=0
#Train accuracy is:  1.0
#Development accuracy is:  0.975
#Test accuracy is:  0.875
#The train accu is very high, but the dev accu is better, test accu is lower

#3rd try (CountVectorizer: ngram_range=(1, 3))
#para: max_iter=5000, random_state=0
#Train accuracy is:  1.0
#Development accuracy is:  0.975
#Test accuracy is:  0.925
#The train accu is very high, but the dev accu is better, test accu is lower

#4th try (CountVectorizer: ngram_range=(2, 3))
#para: max_iter=5000, random_state=0
#Train accuracy is:  1.0
#Development accuracy is:  1.0
#Test accuracy is:  1.0
#The train accu is very high, but the dev accu is better, test accu is ok
clf_lr_udhr = loss_reg(5000, 0,  X_train_udhr, y_train_udhr, X_dev_udhr, y_dev_udhr, X_test_udhr, y_test_udhr)
```

```python
# SVM step

#1st try (TF-IDF features):
#para: random_state=0, tol = 1e-05
#Train accuracy is:  1.0
#Development accuracy is:  1.0
#Test accuracy is:  0.975
#The accu is very high and need to think about if there is data leak or the test set is too small

#2nd try (CountVectorizer: ngram_range=(1, 2))
#para: random_state=0, tol = 1e-05
#Train accuracy is:  1.0
#Development accuracy is:  0.95
#Test accuracy is:  0.9
#The train accu is very high, but the dev accu is better, test accu is ok

#3nd try (CountVectorizer: ngram_range=(1, 2))
#para: random_state=0, tol = 1e-05
#Train accuracy is:  1.0
#Development accuracy is:  0.95
#Test accuracy is:  0.95
#The train accu is very high, but the dev accu is better, test accu is ok

#4th try (CountVectorizer: ngram_range=(1, 3))
#para: random_state=0, tol = 1e-05
#Train accuracy is:  1.0
#Development accuracy is:  0.95
#Test accuracy is:  0.95
#The train accu is very high, but the dev accu is better, test accu is ok

#5th try (CountVectorizer: ngram_range=(2, 3))
#para: random_state=0, tol = 1e-05
#Train accuracy is:  1.0
#Development accuracy is:  1.0
#Test accuracy is:  1.0
#The train, dev, test accu are very high
clf_svm_udhr = svm(0, 1e-05, False, X_train_udhr, y_train_udhr, X_dev_udhr, y_dev_udhr, X_test_udhr, y_test_udhr)
```

```python
#function is to split the text into fragements of n words in dataset
def split_text(testdata, n):
    frag = []
    for _, row in testdata.iterrows():
        words = row["text"].split()
        for i in range(len(words)-n +1):
            f = " ".join(words[i:i+n])
            frag.append({
                "text": f,
                "language": row["language"]
            })
    return pd.DataFrame(frag)
```

```python
#function is to create a new smaller train dataset from big train dataset. 
#Randomly select n texts for each language from the big train dataset
def sample_train_data(data_train, n, random_state=42):
    data_train_en = data_train[data_train["language"] == "en"]
    data_train_nl = data_train[data_train["language"] == "nl"]

    sampled_en = data_train_en.sample(n=n, random_state=random_state)
    sampled_nl = data_train_nl.sample(n=n, random_state=random_state)

    #create new small train dataset
    data_train_sample = pd.concat(
        [sampled_en, sampled_nl],
        ignore_index=True
    )

    data_train_sample = data_train_sample.sample(
        frac=1, random_state=random_state
    ).reset_index(drop=True)

    return data_train_sample
```

```python
#function is to check the learned weights
def check_weights(clf_udhr):
    feature_names = vectorizer.get_feature_names_out()
    weights = clf_udhr.coef_[0]
    
    weight_df = pd.DataFrame({
        "ngram": feature_names,
        "weight": weights
    })
    
    print("Largest positive weights:")
    print(weight_df.sort_values("weight", ascending=False).head(10))
    
    print("Largest negative weights:")
    print(weight_df.sort_values("weight", ascending=True).head(10))
```

```python
#------------- Shorter inputs -------------

#split test text into fragments

#split test text into fragements of 1 words in dataset
data_udhr_test_1frag = split_text(data_udhr_test, 1)
print(data_udhr_test_1frag.head())
print(data_udhr_test_1frag.shape)

#split test text into fragements of 2 words in dataset
data_udhr_test_2frag = split_text(data_udhr_test, 2)
print(data_udhr_test_2frag.head())
print(data_udhr_test_2frag.shape)

#split test text into fragements of 3 words in dataset
data_udhr_test_3frag = split_text(data_udhr_test, 3)
print(data_udhr_test_3frag.head())
print(data_udhr_test_3frag.shape)

#split test text into fragements of 5 words in dataset
data_udhr_test_5frag = split_text(data_udhr_test, 5)
print(data_udhr_test_5frag.head())
print(data_udhr_test_5frag.shape)
```

```python
# text vectorization transform

X_test_udhr_1frag = vectorizer.transform(data_udhr_test_1frag["text"])
X_test_udhr_2frag = vectorizer.transform(data_udhr_test_2frag["text"])
X_test_udhr_3frag = vectorizer.transform(data_udhr_test_3frag["text"])
X_test_udhr_5frag = vectorizer.transform(data_udhr_test_5frag["text"])

y_test_udhr_1frag = data_udhr_test_1frag["language"]
y_test_udhr_2frag = data_udhr_test_2frag["language"]
y_test_udhr_3frag = data_udhr_test_3frag["language"]
y_test_udhr_5frag = data_udhr_test_5frag["language"]

num_frag = [1, 2, 3, 5]
X_test_udhr_fraglist = [X_test_udhr_1frag, X_test_udhr_2frag, X_test_udhr_3frag, X_test_udhr_5frag]
y_test_udhr_fraglist = [y_test_udhr_1frag, y_test_udhr_2frag, y_test_udhr_3frag, y_test_udhr_5frag]
```

```python
# test for Logistic regression

for i in range(len(num_frag)):
    print("Test accuracy for fragements in ", num_frag[i], " words is: ", 
          clf_lr_udhr.score(X_test_udhr_fraglist[i], y_test_udhr_fraglist[i]))
```

```python
# test for SVM

for i in range(len(num_frag)):
    print("Test accuracy for fragements in ", num_frag[i], " words is: ", 
          clf_svm_udhr.score(X_test_udhr_fraglist[i], y_test_udhr_fraglist[i]))
```

```python
#------------- Learning curve -------------

#text vectorization transform

sizes = [1, 2, 5, 10, 20]
X_small = []
y_small = []
for n in sizes:
    data_udhr_train_small = sample_train_data(data_udhr_train, n)

    #just need to vectorize for small train dataset
    X_small.append(vectorizer.transform(data_udhr_train_small["text"]))
    y_small.append(data_udhr_train_small["language"])
```

```python
# test for logistic regression

for i in range(len(sizes)):
    print(f"\nThe train dataset size is {2*sizes[i]}: ")
    clf_lr_udhr = loss_reg(5000, 0,  X_small[i], y_small[i], 
                           X_dev_udhr, y_dev_udhr, X_test_udhr, y_test_udhr)
```

```python
# test for SVM

for i in range(len(sizes)):
    print(f"\nThe train dataset size is {2*sizes[i]}: ")
    clf_svm_udhr = svm(0, 1e-05, False, X_small[i], y_small[i], 
                       X_dev_udhr, y_dev_udhr, X_test_udhr, y_test_udhr)
```
